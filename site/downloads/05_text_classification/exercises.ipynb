## Exercise 1 — Smoothed IDF and normalized rows

A term occurs in 3 of 10 training documents. Derive its scikit-learn smoothed IDF. Compute the IDF of a term in all 10 documents. Explain why smoothing prevents an infinite value when the term is absent from every training document. For a raw weighted vector $(3,4)$, normalize it and verify its norm.

```python
import numpy as np

# TODO: Replace these starter values in your own copy after deriving them.
n_documents = 10
document_frequency = 3
idf = np.log((1 + n_documents) / (1 + document_frequency)) + 1
weighted_vector = np.array([3.0, 4.0])
normalized_vector = weighted_vector / np.linalg.norm(weighted_vector)
print(
    "IDF:",
    idf,
    "normalized vector:",
    normalized_vector,
    "norm:",
    np.linalg.norm(normalized_vector),
)
```

## Exercise 2 — Cosine similarity

For two nonzero documents, show that cosine similarity is the dot product after both vectors have L2 norm one. Calculate the cosine similarity of $(1,0,1)$ and $(1,1,0)$. Why does a normalized sparse representation make the calculation efficient?

```python
u = np.array([1.0, 0.0, 1.0])
v = np.array([1.0, 1.0, 0.0])
# TODO: replace the placeholder with your cosine calculation.
cosine = np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))
print("cosine:", cosine)
```

## Exercise 3 — Softmax cross-entropy derivative

For one example with $K$ logits, softmax probabilities $p_k$, and one-hot target $y_k$, derive $\partial[-\sum_r y_r\log p_r]/\partial z_k=p_k-y_k$. Then state the batch gradient with feature vector $x_i$. Explain in words how an L2 penalty changes the fitted weights.

```python
# Numerical check for one target class (category index 1).
logits = np.array([0.2, -0.1, 0.7])
probabilities = np.exp(logits - logits.max())
probabilities /= probabilities.sum()
y = np.array([0.0, 1.0, 0.0])
analytic_gradient = probabilities - y
print("probabilities:", probabilities)
print("softmax-cross-entropy gradient with respect to logits:", analytic_gradient)
```
